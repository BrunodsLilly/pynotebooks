```python
import jax.numpy as jnp
from typing import Any, Callable
```

```python
#| export
def cost(y: jnp.ndarray, f: Callable[..., jnp.ndarray], *params: Any) -> float:
    """Compute the cost function for linear regression.
    
    Parameters
    ----------
    y : ndarray of shape (m,)
        Target values for the training examples.
    f : callable
        Prediction function that takes parameters and returns predictions.
    *params
        Variable length argument list of parameters to be passed to the 
        prediction function f.
    
    Returns
    -------
    float
        The mean squared error cost of using the given parameters for 
        linear regression to fit the data points.
    """
    m = len(y)  # the number of training examples
    errors = f(*params) - y
    
    # return the sum of squared errors
    return 1/(2*m) * jnp.dot(errors, errors)
```

```python
x_train = jnp.array([1.0, 1.7, 2.0, 2.5, 3.0, 3.2])
y_train = jnp.array([250, 300, 480,  430,   630, 730,])
cost(y_train, lambda x, w, b: jnp.dot(x, w) + b, x_train, 200, -100)
```
